```python
# rows where total_bill > 20
df[df["total_bill"] > 20]
```

```python
#Sort — the whole dataset by tip, descending.
df.sort_values(by="tip", ascending=False)
```

```python
#Groupby — average tip for each day.
df.groupby("day")["tip"].mean()
```

```python
#find discrete values in all column
df.nunique()
```

```python
#creating a small second DataFrame mapping day abbreviations to full names (e.g. {"Thur": "Thursday", "Fri": "Friday", ...} then merge it onto df using day as the key.
day_df = pd.DataFrame({"day": ["Thur", "Fri", "Sat", "Sun"], "full_day": ["Thursday", "Friday", "Saturday", "Sunday"]})
df.merge(day_df, on="day")
```

```python
df
```

```python
#a pivot table of average total_bill by day (rows) and time (columns), using pd.pivot_table
df.pivot_table(values="total_bill", index="day", columns="time")
```

```python
df.pivot_table(values="total_bill", index="day", columns="time", aggfunc=["sum","min","max","count"])
```
